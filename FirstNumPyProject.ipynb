```python
import numpy as np

def generate_weather_data():
    #EXPLORING WEATHER DATA USING NumPy
    np.random.seed(42)
    days = np.arange(1, 366)
    min_temps = np.random.randint(-10, 16, size=365)
    max_temps = np.random.randint(5, 36, size=365)
    return np.column_stack((days, min_temps, max_temps))
    raise NotImplementedError()
```

```python
def basic_statistics(data):
    #Calculate basic statistics using the data from generate_weather_data() funtion
    days = data[:, 0]
    min_temps = data[:, 1]
    max_temps = data[:, 2]
    
    avg_min_temp = np.mean(min_temps)
    avg_max_temp = np.mean(max_temps)
    
    high_temp = np.max(max_temps)
    high_day = days[np.argmax(max_temps)]
    
    low_temp = np.min(min_temps)
    low_day = days[np.argmin(min_temps)]
    result = avg_min_temp, avg_max_temp, (high_temp,high_day),(low_temp,low_day)
    return result
    raise NotImplementedError()
```

```python
def daily_temperature_range(data):
    # Function computes the daily temperature range (maximum - minimum) for all days
    temp_ranges = data[:, 2] - data[:, 1]
    max_range_day = np.argmax(temp_ranges) + 1
    max_range_value = np.max(temp_ranges)
    return temp_ranges, (max_range_day, max_range_value)
    raise NotImplementedError()
```

```python
def identify_heatwaves(data):
    days = data[:, 0].astype(int)
    max_temps = data[:, 2]  # Column 3: max temperature

    # Boolean array: True where max temp > 30
    hot = max_temps > 30

    heatwaves = []
    i = 0
    n = len(hot)

    while i < n:
        if hot[i]:
            start = i
            while i < n and hot[i]:
                i += 1
            end = i - 1
            if (end - start + 1) >= 3:
                heatwaves.append((int(days[start]), int(days[end])))
        else:
            i += 1

    return len(heatwaves), heatwaves
    raise NotImplementedError()

#Test cases to validate the code
data = generate_weather_data()

# Test Case 1: Dataset generation
print(data.shape)

# Test Case 2: Basic statistics
avg_min, avg_max, max_temp, min_temp = basic_statistics(data)
assert np.isclose(avg_min, 2.10, atol=0.01),                             "Test Case 2 Failed: Incorrect average minimum temperature."
assert np.isclose(avg_max, 20.89, atol=0.01),                            "Test Case 2 Failed: Incorrect average maximum temperature."

# Test Case 3: Daily temperature range
ranges, max_range_info = daily_temperature_range(data)
print(max_range_info)
```
