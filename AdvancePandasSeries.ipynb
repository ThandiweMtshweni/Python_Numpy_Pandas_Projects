```python
import pandas as pd

def access_using_loc_iloc(series, loc_index, iloc_index):
    #Accessing Series Objects using loc and iloc Indexers
    loc_value = int(series.loc[loc_index])
    iloc_value = int(series.iloc[iloc_index])
    return loc_value, iloc_value
    raise NotImplementedError()
s = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
print(access_using_loc_iloc(s, 'b', 2))
print(access_using_loc_iloc(s, 'a', 0))
print(access_using_loc_iloc(s, 'c', 1))
```

```python
def access_with_iloc(series, idx):
    # YOUR CODE HERE
    return series.iloc[idx]
    raise NotImplementedError()
s = pd.Series([10, 20, 30])
print(access_with_iloc(s, 1))
print(access_with_iloc(s, 0))
print(access_with_iloc(s, 2))
```

```python
def access_with_iat(series, idx):
    # YOUR CODE HERE
    return series.iat[idx]
    raise NotImplementedError()
s = pd.Series([10, 20, 30])
print(access_with_iat(s, 2))
print(access_with_iat(s, 0))
print(access_with_iat(s, 1))
```

```python

def series_attributes(series):
    # Series 
    return {
        "array" : series.array,
        "axes" : series.axes,
        "dtypes" : series.dtype,
        "hasnans" : series.hasnans,
        "index" : series.index,
        "is_unique" : series.is_unique,
        "nbytes" : series.nbytes,
        "size" : series.size,
        "values" : series.values
    }
    raise NotImplementedError()
s = pd.Series([10, 20, 30])
attrs = series_attributes(s)
print(attrs['size'])
print(not attrs['hasnans'])
print(attrs['is_unique'])
```
