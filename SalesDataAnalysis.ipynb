```python
def handle_missing_values():
    data = {
        "OrderID": [1001, 1002, 1003, 1004, 1005],
        "Product": ["Laptop", None, "Office Chair", "Desk", "Microwave"],
        "Category": ["Electronics", "Electronics", "Furniture", "Furniture", None],
        "Customer": ["Alice Smith", "Bob Johnson", "Charlie Lee", None, "Eve Wilson"],
        "Quantity": [2, 1, 5, None, 1],
        "Price": [800, 600, 120, 200, None],
        "Discount": [10, 5, None, 0, 10],
        "Total": [1440, None, 510, 200, 135],
        "Region": ["North", "South", "East", None, "North"],
        "Date": ["2023-01-15", "2023-01-18", "2023-01-20", "2023-02-01", "2023-02-15"]}
    
    df = pd.DataFrame(data)
    missing_counts = df.isna().sum()

    df["Product"] = df["Product"].fillna("Unknown")
    df["Category"] = df["Category"].fillna("Unknown")
    df["Customer"] = df["Customer"].fillna("Not Specified")
    df["Region"] = df["Region"].fillna("Not Specified")

    df["Quantity"] = df["Quantity"].fillna(0)
    df["Price"] = df["Price"].fillna(0)

    df["Discount"] = df["Discount"].fillna(df["Discount"].mean())
    df["Total"]    = df["Total"].fillna(df["Total"].mean())

    return missing_counts, df

missing_counts, modified_df = handle_missing_values()
isinstance(missing_counts, pd.Series)
isinstance(modified_df, pd.DataFrame)
print(missing_counts["Product"])
print(modified_df.loc[1, "Product"])
print(modified_df.loc[3, "Customer"])
```

```python
def aggregate_and_group():
    # YOUR CODE HERE
    data = {
    "Region": ["North", "South", "East", "West", "North", "South"],
    "Category": ["Electronics", "Electronics", "Furniture", "Furniture", "Appliances", "Appliances"],
    "Sales": [1200, 800, 950, 650, 400, 300],
    "Profit": [200, 150, 120, 50, 60, 40]
    }

    df = pd.DataFrame(data)

    result = (
        df.groupby(["Region", "Category"])
          .agg(
              TotalSales=("Sales", "sum"),
              AvgProfit=("Profit", "mean")
          )
          .reset_index()
    )

    return result
    raise NotImplementedError()
result = aggregate_and_group()

isinstance(result, pd.DataFrame),
"TotalSales" in result.columns,
print(result.loc[(result["Region"] == "North") & (result["Category"] == "Electronics"), "TotalSales"].values[0])
print(result.loc[(result["Region"] == "South") & (result["Category"] == "Appliances"), "AvgProfit"].values[0])
```

```python

def create_sales_pivot_table():
    # Function 
    data = {
    "Region": ["North", "South", "East", "West", "North", "South", "East", "West"],
    "Category": ["Electronics", "Electronics", "Furniture", "Furniture", "Appliances", "Appliances", "Electronics", "Furniture"],
    "Sales": [1200, 800, 950, 650, 400, 300, 1500, 700],
    "Profit": [200, 150, 120, 50, 60, 40, 300, 100]
    }

    df = pd.DataFrame(data)

    result = pd.pivot_table(
        df,
        index="Region",
        columns="Category",
        values="Sales",
        aggfunc="sum",
        fill_value=0
    )

    return result
    raise NotImplementedError()

result = create_sales_pivot_table()

isinstance(result, pd.DataFrame),
print(result.loc["North", "Electronics"])
print(result.loc["South", "Appliances"])
print(result.loc["West", "Furniture"])
```
