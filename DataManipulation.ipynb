```python
import pandas as pd
products = [
    ("Laptop", 1000),
    ("Phone", 500),
    ("Headphones", 100),
    ("Mouse", 50),
    ("Monitor", 200),
    ("Keyboard", 150)
]

def filter_products(products, min_price, max_price):
    """
    Filters products within a specified price range.

    Parameters:
        products (list): A list of tuples, where each tuple contains the product name (str) and price (float).
        min_price (float): The minimum price in the range.
        max_price (float): The maximum price in the range.

    Returns:
        list: A list of product names within the specified price range.
    """
    # YOUR CODE HERE
    return [name for name, price in products
        if min_price <= price <= max_price
    ]
    raise NotImplementedError()
print(filter_products(products, 100, 700))
print(filter_products(products, 0, 100))
print(filter_products(products, 700, 1500))
```

```python
orders = [
    {"order_id": 1, "region": "North"},
    {"order_id": 2, "region": "South"},
    {"order_id": 3, "region": "North"},
    {"order_id": 4, "region": "East"},
    {"order_id": 5, "region": "South"},
    {"order_id": 6, "region": "West"}
]

def count_orders_by_region(orders):
    """
    Counts the number of orders in each region.

    Parameters:
        orders (list): A list of dictionaries, where each dictionary contains 'order_id' (int) and 'region' (str).

    Returns:
        dict: A dictionary where keys are region names and values are the count of orders.
    """
    # YOUR CODE HERE
    region_count = {}
    for orders in orders:
        region = orders["region"]
        region_count[region] = region_count.get(region, 0) + 1
    return region_count
    raise NotImplementedError()
print(count_orders_by_region(orders))
print(count_orders_by_region([
    {"order_id": 1, "region": "North"},
    {"order_id": 2, "region": "North"}
]))
print(count_orders_by_region([]))
```

```python
import numpy as np

sales = np.array([
    [5, 20.0],  # 5 units sold at $20 each
    [10, 15.0], # 10 units sold at $15 each
    [3, 50.0],  # 3 units sold at $50 each
    [8, 10.0]   # 8 units sold at $10 each
])

def sales_summary(sales):
    """
    Calculates the total and average revenue from sales data.

    Parameters:
        sales (np.ndarray): A NumPy array where each row contains [units_sold, unit_price].

    Returns:
        tuple: A tuple containing:
            - total_revenue (float): The total revenue.
            - average_revenue (float): The average revenue per sale, rounded to 2 decimal places.
    """
    # YOUR CODE HERE
    revenue = sales[:,0] * sales[:,1]
    total_revenue = revenue.sum()
    average_revenue = revenue.mean()
    return total_revenue, average_revenue
    raise NotImplementedError()
sales = np.array([
    [5, 20.0],
    [10, 15.0],
    [3, 50.0],
    [8, 10.0]
])
print(sales_summary(sales))
sales_test = np.array([
    [2, 100.0],
    [4, 50.0],
    [1, 200.0]
])
print(sales_summary(sales_test))
sales_test_empty = np.array([[0, 0]])
print(sales_summary(sales_test_empty))
```

```python
import pandas as pd

def apply_discounts(prices):
    """
    Applies a 10% discount to prices greater than 100.

    Parameters:
        prices (pd.Series): A Pandas Series containing product prices.

    Returns:
        pd.Series: A new Series with the updated prices.
    """
    # YOUR CODE HERE
    discounted_prices = prices.astype(float).copy()
    discounted_prices[discounted_prices > 100] *= 0.9
    return discounted_prices
    raise NotImplementedError()
prices = pd.Series([50, 150, 200, 80, 120])
updated_prices = apply_discounts(prices)
print(updated_prices.tolist())

prices_test = pd.Series([100, 101, 99, 200, 150])
updated_prices_test = apply_discounts(prices_test)
print(updated_prices_test.tolist())

prices_empty = pd.Series([])
updated_prices_empty = apply_discounts(prices_empty)
print(updated_prices_empty.tolist())
```

```python
import pandas as pd

def analyze_orders(orders):
    """
    Analyzes customer orders to calculate total revenue and the highest value order.

    Parameters:
        orders (pd.DataFrame): A DataFrame containing columns 'OrderID', 'Customer', 'Quantity', and 'UnitPrice'.

    Returns:
        dict: A dictionary containing:
            - 'total_revenue' (float): The total revenue from all orders.
            - 'highest_order' (dict): Details of the highest value order.
    """
    # YOUR CODE HERE
    df = orders.copy()
    df["OrderValue"] = df["Quantity"] * df["UnitPrice"]
    total_revenue = float(df["OrderValue"].sum())
    highest_row = df.loc[df["OrderValue"].idxmax()]
    highest_order = {
        "OrderID": int(highest_row["OrderID"]),
        "Customer": highest_row["Customer"],
        "OrderValue": float(highest_row["OrderValue"])
    }

    return {
        "total_revenue": total_revenue,
        "highest_order": highest_order
    }
    raise NotImplementedError()

orders = pd.DataFrame({
    "OrderID": [1, 2, 3, 4, 5],
    "Customer": ["Alice", "Bob", "Charlie", "David", "Eve"],
    "Quantity": [2, 1, 3, 5, 2],
    "UnitPrice": [20.0, 50.0, 15.0, 10.0, 100.0]
})
result = analyze_orders(orders)
print(result["total_revenue"])
print(result["highest_order"])
orders_test = pd.DataFrame({
    "OrderID": [6, 7, 8],
    "Customer": ["Frank", "Grace", "Hank"],
    "Quantity": [4, 2, 6],
    "UnitPrice": [30.0, 25.0, 40.0]
})
result_test = analyze_orders(orders_test)
print(result_test["total_revenue"])
print(result_test["highest_order"])
```

```python
from collections import Counter
import string
import re

def parse_feedback(feedback):
    """
    Parses customer feedback to extract word count, most common word, and cleaned feedback.

    Parameters:
        feedback (str): The customer feedback string.

    Returns:
        dict: A dictionary containing:
            - "word_count" (int): Total number of words in the feedback.
            - "most_common_word" (str): The word that appears most frequently.
            - "cleaned_feedback" (str): The cleaned version of the feedback.
    """
    cleaned = feedback.lower()
    words = cleaned.split()
    word_count = len(words)
    most_common_word = Counter(words).most_common(1)[0][0] if words else None
    words = cleaned.split()
    return {"word_count": word_count,
     "most_common_word": most_common_word,
     "cleaned_feedback" : words}
    # YOUR CODE HERE
    raise NotImplementedError()
feedback = "Great service! Product arrived on time. The packaging was excellent, and the delivery was fast. Great service!"
result = parse_feedback(feedback)
print(result["word_count"])
print(result["most_common_word"])
print(result["cleaned_feedback"])
```
