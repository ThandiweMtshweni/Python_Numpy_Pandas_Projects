```python
import pandas as pd
def find_outliers(series):
    #Identifying Outliers Using the IQR Method
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    is_outlier = (series < lower_bound) | (series > upper_bound)
    return is_outlier
    raise NotImplementedError()
s = pd.Series([10, 12, 15, 10, 300])
result = find_outliers(s)
print(result.tolist())
s = pd.Series([1, 2, 3, 4, 100])
result = find_outliers(s)
print(result.tolist())
s = pd.Series([10, 20, 30, 40])
result = find_outliers(s)
print(result.tolist())
```

```python
def fill_missing_with_mean(series):
    #Filling Missing Data with Logical Rules
    if series.isna().all():
        return series
    else:
        mean_value = series.mean(skipna=True)
        new_values = series.fillna(mean_value)
        return new_values
    raise NotImplementedError()
s = pd.Series([1, None, 3, 4])
result = fill_missing_with_mean(s)
print(result.tolist())

s = pd.Series([10, None, 30, None])
result = fill_missing_with_mean(s)
print(result.tolist())

s = pd.Series([None, None, None])
result = fill_missing_with_mean(s)
print(result.tolist())
```

```python
def grouped_mean(series, level):
    # Multi-Indexed Series – Aggregation by Groups
    return series.groupby(level=level).mean()
    raise NotImplementedError()
data = {('A', 1): 10, ('A', 2): 20, ('B', 1): 30, ('B', 2): 40}
s = pd.Series(data)
result = grouped_mean(s, level=0)
print(result.tolist())

result = grouped_mean(s, level=1)
print(result.tolist())
```
